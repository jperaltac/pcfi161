```python
### ------- Pregunta 1 --- --------
import numpy as np 
import matplotlib.pyplot as plt
#primero debemos definir las constantes
b1 = 6.5
b2 = 2.8
b3 = 1.4
q = 90
a1 = np.pi / 4
a2 = -np.pi / 6

#creamos el arreglo con linspace 
t = np.linspace(0, 180, 400)

#y luego creamos la funcion
def temperatura_panel(t):
    T = 22 + b1 + np.sin(2 * np.pi * t / q) + b2 * np.cos(4 * np.pi * t / q + a1) + b3 * np.sin(6 * np.pi * t / q + a2)
    return T

temp_panel = temperatura_panel(t)

#hacemos el grafico con plot uwu
plt.plot(t, temp_panel, color='indigo', label='vacioacion de la temperatura')
plt.title('Panel de temperatura')
plt.xlabel('Tiempo')
plt.ylabel('Temperatura')
plt.legend()
plt.grid(True)
plt.show()

#luego debemos calcular las temperaturas
t_max = np.max(temp_panel)
print(f'la temperatura maxima es: {t_max}')
t_min = np.min(temp_panel)
print(f'la temperatura minima es: {t_min}')
t_prom = np.mean(temp_panel)
print(f'la temperatura promedio es: {t_prom}')

#creamos una variable que tenga la amplitud termica
At = (t_max - t_min) / 2
print(f'la amplitud termica es: {At}')

#creamos los condicionales para evaluar si es intensa o moderada
Rt = At / t_prom
if Rt >= 0.20:
    print(f'la variacion termica es intensa')
else:
    print(f'la variacion termica es moderada')
```

```python
# ------- pregunta 2 ------------
import numpy as np
import matplotlib.pyplot as plt

#primero definimos las variables constantes
h0 = 80
vx0 = 12
vy0 = 0
aw = 0.8
g = 9.8

#creamos el arreglo temporal
t = np.linspace(0, 5, 300)

#creamos las variables para calcular las ecuaciones cinematicas uwu
#primero los valores de posicion
xt = vx0 * t + (1 / 2) * aw * t ** 2
yt = h0 + vy0 * t - (1 / 2) * g * t**2

#luego los valores de velocidad
vxt = vx0 + (aw * t) 
vyt = vy0 - (g * t)

#encontramos el indice y valor minimo de la posicion vertical
pos_ind = np.mean(yt)
print(f'el indice de la posicion es: {pos_ind}')
val_min = np.min(yt)
print(f'el valor minimo de la posicion vertical es: {val_min}')

#calculamos los promedios de la velocidad
prom_vx = np.mean(vxt)
print(f'el promedio de la velocidad es: {prom_vx}')
prom_vy = np.mean(vyt)

#creamos una variable para calcular los valores de aceleracion
unos = np.ones(300)
axt = aw * unos
ayt = -g * unos 


#luego creamos el grafico con fig ax
fig, ax = plt.subplots(3,2, figsize=(12,7))

ax[0,0].plot(t, yt, color='darkblue', label='movimiento vertical')
ax[0,0].set_title('Movimiento vertical')
ax[0,0].set_xlabel('tiempo')
ax[0,0].set_ylabel('posicion vertical')
ax[0,0].grid(True)
ax[0,0].legend()

ax[1,0].plot(t, vyt, color='purple', label='velocidad')
ax[1,0].set_title('Velocidad en y')
ax[1,0].set_xlabel('tiempo')
ax[1,0].set_ylabel('velocidad en direccion y')
ax[1,0].grid(True)
ax[1,0].legend()

ax[2,0].plot(t, ayt, color='darkmagenta', label='aceleracion g')
ax[2,0].set_title('Aceleracion')
ax[2,0].set_xlabel('tiempo')
ax[2,0].set_ylabel('aceleracion')
ax[2,0].grid(True)
ax[2,0].legend()

ax[0,1].plot(t, xt, color='purple', label='posiciones')
ax[0,1].set_title('posicion en x')
ax[0,1].set_xlabel('tiempo')
ax[0,1].set_ylabel('posiciones en x')
ax[0,1].grid(True)
ax[0,1].legend()

ax[1,1].plot(t, vxt, color='navy', label='velocidad')
ax[1,1].set_title('Velocidad en x')
ax[1,1].set_xlabel('tiempo')
ax[1,1].set_ylabel('velocidad en x')
ax[1,1].grid(True)
ax[1,1].legend()

ax[2,1].plot(t, axt, color='cyan', label='aceleracion en x')
ax[2,1].set_title('Aceleracion')
ax[2,1].set_xlabel('tiempo')
ax[2,1].set_ylabel('aceleracion en x')
ax[2,1].grid(True)
ax[2,1].legend()
```

```python
import numpy as np
import matplotlib.pyplot as plt

#primero creamos los arreglos de luminosidad y distancia orbital
L = np.array([1.2, 0.8, 2.5, 1.6, 0.5, 3.0, 1.1])
d = np.array([1.0, 0.7, 1.8, 1.2, 0.4, 2.5, 1.6])

#creamos una lista vacia que contenga el valor de S
S = []

#creamos un bucle de for para calcular y que recorra todos los valores de L y d .-.
for i in range(7):
    Si = L[i] / d[i] **2

    S.append(Si)

#luego creamos los condiciomales para evaluar el flujo 
if Si > 2:
    print('hay flujo alto')
elif 0.8 < Si <= 2:
    print('hay un flujo intermedio')
elif Si <= 0.8:
    print('hay flujo bajo')

#debemos mostrar el indice, la luminosidad, distancia orbital 
print(f'indice del exoplaneta es: {L[i]}')
print(f'la distancia orbital es: {d[i]}')
print(f'el flujo calculado es: {Si}')

#calculamos el flujo promedio de los exoplanetas
flujo_prom = np.mean(Si)
print(f'el flujo promedio es: {flujo_prom}')

#y luego creamos el grafico
plt.scatter(d, S, color='blueviolet', label='flujo recibido')
plt.xlabel('distancia')
plt.ylabel('flujo recibido')
plt.legend()
plt.grid(True)
plt.show()
```
