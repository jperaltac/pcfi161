## Verificar estado de archivo PDF

```python
# Verificar el estado del archivo PDF con ID específico
file_id = 18144690

# Obtener el curso actual
course = Canvas_Key.canvas.get_course(Canvas_Key.current_course_id)

# Obtener el archivo
pdf_file = course.get_file(file_id)

print("=" * 60)
print(f"📄 INFORMACIÓN DEL ARCHIVO PDF")
print("=" * 60)
print(f"ID: {pdf_file.id}")
print(f"Nombre: {pdf_file.display_name}")
print(f"Tamaño: {pdf_file.size / 1024:.2f} KB")
print(f"URL: {pdf_file.url}")
print()
print("🔒 CONFIGURACIÓN DE PRIVACIDAD:")
print(f"Hidden (oculto): {pdf_file.hidden}")
print(f"Locked (bloqueado): {pdf_file.locked}")
print(f"Visibility: {getattr(pdf_file, 'visibility', 'N/A')}")
print()
print("📅 FECHAS DE DISPONIBILIDAD:")
print(f"Unlock at: {getattr(pdf_file, 'unlock_at', 'N/A')}")
print(f"Lock at: {getattr(pdf_file, 'lock_at', 'N/A')}")
print("=" * 60)
```
