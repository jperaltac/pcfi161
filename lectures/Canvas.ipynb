```python
# Buscar y listar los items del módulo llamado "Material del curso"
target_name = "Material del curso"
found = None

for m in course.get_modules():
    if m.name and m.name.strip().lower() == target_name.lower():
        found = m
        break

if not found:
    for m in course.get_modules():
        if m.name and target_name.lower() in m.name.strip().lower():
            found = m
            break

if not found:
    print("No se encontró un módulo con nombre 'Material del curso'. Módulos disponibles:")
    for m in course.get_modules():
        print(f"{m.id}: {m.name} (published={m.published})")
else:
    print(f"Encontrado módulo: {found.id}: {found.name} (published={found.published})\n")
    print("=" * 80)
    
    # Intentar distintos métodos para obtener los items del módulo
    try:
        items = found.get_module_items()
    except Exception:
        try:
            items = found.get_items()
        except Exception:
            items = course.get_module(found.id).get_module_items()

    # Mostrar propiedades detalladas de cada item, especialmente SubHeaders
    for it in items:
        it_type = getattr(it, "type", getattr(it, "module_item_type", ""))
        it_title = getattr(it, "title", getattr(it, "name", ""))
        
        print(f"\n📌 Item ID: {it.id}")
        print(f"   Type: {it_type}")
        print(f"   Title: {it_title}")
        
        # Si es SubHeader, mostrar TODAS sus propiedades
        if it_type == "SubHeader":
            print(f"   🔍 PROPIEDADES DE SUBHEADER:")
            print(f"   " + "-" * 60)
            
            # Listar todos los atributos del objeto
            for attr in dir(it):
                # Ignorar métodos privados y métodos callable
                if not attr.startswith('_') and not callable(getattr(it, attr, None)):
                    try:
                        value = getattr(it, attr)
                        print(f"   {attr}: {value}")
                    except Exception as e:
                        print(f"   {attr}: <Error al acceder: {e}>")
        else:
            # Para otros tipos, mostrar info básica
            print(f"   Published: {getattr(it, 'published', 'N/A')}")
            print(f"   Indent: {getattr(it, 'indent', 'N/A')}")
            print(f"   Position: {getattr(it, 'position', 'N/A')}")
        
        print("   " + "-" * 60)
```
