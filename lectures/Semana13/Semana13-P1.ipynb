## Bubble sort

[web] https://takeuforward.org/arrays/recursive-bubble-sort-algorithm/

[video] https://youtu.be/kPRA0W1kECg?t=242

```python
import pandas as pd 
pd.read_csv('https://raw.githubusercontent.com/jperaltac/pcfi161/main/tareas/tarea03/estrellas.csv')
```

```python
def bubble_sort(lista):
    n=len(lista)
    for i in range(n-1):
        for j in range(n-1-i):
            if lista[j] > lista[j+1]:
                lista[j], lista[j+1] = lista[j+1], lista[j]
    return lista
```

```python
n = 4

for i in range(n-1):
    for j in range(n-1-i):
        print(j,j+1)
```

## Ejemplo – Ordenar distancias planetarias

```python
dist_au = [1.52, 0.39, 5.20, 1.00, 30.07]
print("Original: ", dist_au)
print("Ordenado: ", bubble_sort(dist_au))
```

```python
def linear_search(lista, objetivo):
    for i, x in enumerate(lista):
        if x == objetivo:
            return i
    return -1 #si no existe retorna -1
```

## Binary search (en lista ordenada)

```python
def binary_search(lista_ordenada, objetivo): # Solo funciona si la lista esta ordenada
    low, high = 0, len(lista_ordenada)-1
    while low <= high:
        mid = (low + high)//2
        if lista_ordenada[mid] == objetivo:
            return mid
        elif lista_ordenada[mid] < objetivo:
            low = mid + 1
        else:
            low = mid - 1
    return -1
```

## Ejercicio 2 - Búsqueda Binaria

```python
import numpy as np
mags = np.random.normal(8, 1.5, size=300).round(2)
```

```python
mags_ordenadas = (mags)
```

```python
print(mags_ordenadas)
```

### Paso 2 - Definición de la funciónde búsqueda binaria

```python
def binary_search(lst, target): #similar al anterior pero retorna cuantos 'checks' ha realizado
    low, high, checks = 0, len(lst)-1, 0
    while low <= high:
        mid = (low + high) //2
        checks += 1
        if lst[mid] == target:
            return mid, checks
        elif target < lst[mid]:
            high = mid - 1
        else:
            low = mid + 1
    return -1, checks
```

### Paso 3 - Ejecución y comparación

```python
val = float(input("Ingresa magnitud a buscar: " )) #Poner el último valor de la lista ayuda para la idea de cual tarda más.
pos, nchecks = binary_search(mags_ordenadas, val)
print(f"La magnitud {val} se encuentra en la posición {pos} y se realizaron {nchecks} checks")
```

```python
def linear_search(lst, target): #Similar al de antes, pero retorna los 'checks'
    for checks, x in enumerate(lst, 1):
        if x == target:
            return checks
    return checks #no encoentrado, barrió toda la lista.
```

```python
binary_search(mags_ordenadas.copy(), val)
```

```python
%timeit binary_search(mags_ordenadas, val)
%timeit linear_search(mags_ordenadas, val)
```
