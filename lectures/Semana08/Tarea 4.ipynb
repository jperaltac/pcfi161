```python
import pandas as pd
import numpy as np
```

```python
# Cargar el archivo titanic.csv desde la URL pública
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
df
```

```python
print("shape:", df.shape)
```

```python
df.info()
```

```python
df.describe()
```

```python
#Calcular el \% de supervivencia general
survival_rate = df['Survived'].mean()*100
survival_rate
```

```python
# Calcular el \% de supervivencia por clase
survival_by_class = df.groupby('Pclass')['Survived'].mean().plot(kind='bar')
survival_by_class
```
