```python
import numpy as np

# a) Resolver h(t) = 0 usando np.roots
coeficientes = [-4.9, 18, 1.5]
raices = np.roots(coeficientes)

# b) Filtrar raíces reales
raices_reales = []

for r in raices:
    # r.imag entrega la parte imaginaria de la raíz r.
    # Por ejemplo, si r = 2 + 0j, entonces r.imag = 0.
    # np.isclose compara si r.imag es prácticamente igual a 0,
    # considerando pequeños errores numéricos de aproximación.
    if np.isclose(r.imag, 0):
        raices_reales.append(r.real)

# c) Identificar cuál raíz tiene sentido físico
for t in raices_reales:
    if t > 0:
        tiempo_impacto = t

# d) Interpretar el resultado como tiempo de impacto con el suelo
print("El tiempo de impacto con el suelo es:", tiempo_impacto, "s")
```
